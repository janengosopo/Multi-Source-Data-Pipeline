```python
import pandas as pd
products_df = pd.read_csv(
    "/Volumes/workspace/sales/sales/products.csv",
    keep_default_na=False,
    na_values=[""],
)
```

```python
# Quick look at what we're dealing with
print("Shape:", products_df.shape)
print(products_df.dtypes)
print("\nMissing values:")
print(products_df.isna().sum())
print("\nDuplicates:", products_df.duplicated().sum())
print("\nSample of category values:", products_df["category"].unique())
print("Sample of feedstock values:", products_df["feedstock"].unique())
```

```python
products_df
```

```python
# --- Clean product_name
products_df["product_name"] = products_df["product_name"].str.strip()
```

```python
products_df["product_name"].unique()
```

```python
# --- Clean category
products_df["category"] = products_df["category"].str.strip().str.title()
```

```python
products_df["category"].unique()
```

```python
# --- Clean feedstock
products_df["feedstock"] = products_df["feedstock"].str.strip().str.title()
```

```python
# --- Clean standard_cost_eur_per_mt
products_df["standard_cost_eur_per_mt"] = pd.to_numeric(
    products_df["standard_cost_eur_per_mt"]
        .astype(str)
        .str.replace("€", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip(),
    errors="coerce",
)
```
