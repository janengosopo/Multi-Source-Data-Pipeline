```python
import pandas as pd
# Note: keep_default_na=False prevents pandas from turning "NA" (North America)
customers_df = pd.read_csv(
    "/Volumes/workspace/sales/sales/customers.csv",
    keep_default_na=False,
    na_values=[""],   # only treat truly empty cells as missing
)
```

```python
# Quick look at what we're dealing with
print("Shape:", customers_df.shape)
print(customers_df.dtypes)
print(customers_df.isna().sum())
print("\nDuplicates:", customers_df.duplicated().sum())
print("\nSample of region values:", customers_df["region"].unique())
print("Sample of industry values:", customers_df["industry"].unique())
```

```python
customers_df
```

```python
customers_df["region"].unique()
```

```python
# --- Clean customer_name
customers_df["customer_name"] = customers_df["customer_name"].str.strip()
```

```python
customers_df["customer_name"]
```

```python
# --- Clean industry
customers_df["industry"] = customers_df["industry"].str.strip().str.title()
```

```python
customers_df["industry"].unique()
```

```python
# --- Clean credit_limit_eur
customers_df["credit_limit_eur"] = pd.to_numeric(
    customers_df["credit_limit_eur"]
        .astype(str)
        .str.replace("€", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip(),
    errors="coerce",
)
```

```python
customers_df["credit_limit_eur"]
```

```python
# --- Clean payment_terms_days
customers_df["payment_terms_days"] = pd.to_numeric(
    customers_df["payment_terms_days"], errors="coerce"
)
```

```python
customers_df["payment_terms_days"]
```

```python
# --- Drop duplicate rows 
before = len(customers_df)
customers_df = customers_df.drop_duplicates().reset_index(drop=True)
print(f"\nRemoved {before - len(customers_df)} duplicate row(s)")
```
